```python
import requests
import pandas as pd

# NEXORA SCIENCE — Récupération de données météorologiques

url = "https://power.larc.nasa.gov/api/temporal/daily/point"

parametres = {
    "parameters": "T2M",
    "community": "AG",
    "longitude": -17.45,
    "latitude": 14.69,
    "start": "20240101",
    "end": "20241231",
    "format": "JSON"
}

reponse = requests.get(url, params=parametres, timeout=60)
reponse.raise_for_status()

donnees = reponse.json()["properties"]["parameter"]["T2M"]

df = pd.DataFrame(
    list(donnees.items()),
    columns=["Date", "Temperature_C"]
)

df["Date"] = pd.to_datetime(df["Date"], format="%Y%m%d")
df["Temperature_C"] = pd.to_numeric(
    df["Temperature_C"], errors="coerce"
)

df = df.dropna().reset_index(drop=True)

print("NEXORA SCIENCE — Données météorologiques de Dakar")
print("Nombre de mesures :", len(df))
print("\nPremières mesures :")
print(df.head())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# NEXORA SCIENCE — Modèle expérimental
# Utilisation des températures historiques de Dakar

# Préparer les données
df_modele = df[["Date", "Temperature_C"]].dropna().copy()
df_modele = df_modele.sort_values("Date").reset_index(drop=True)

# Utiliser les 3 jours précédents pour prédire le jour suivant
X = []
y = []

for i in range(3, len(df_modele)):
    X.append(df_modele["Temperature_C"].iloc[i-3:i].values)
    y.append(df_modele["Temperature_C"].iloc[i])

X = np.array(X)
y = np.array(y)

# Séparation chronologique : premières 80 % pour l'entraînement,
# dernières 20 % pour le test
separation = int(len(X) * 0.8)

X_train, X_test = X[:separation], X[separation:]
y_train, y_test = y[:separation], y[separation:]

# Entraîner le modèle
modele = LinearRegression()
modele.fit(X_train, y_train)

# Faire les prédictions
predictions = modele.predict(X_test)

# Évaluer les résultats
mae = mean_absolute_error(y_test, predictions)

print("NEXORA SCIENCE — Rapport expérimental")
print("Nombre de données d'entraînement :", len(X_train))
print("Nombre de données de test :", len(X_test))
print("Erreur moyenne absolue :", round(mae, 2), "°C")

# Visualiser les résultats
dates_test = df_modele["Date"].iloc[3 + separation:]

plt.figure(figsize=(12, 5))
plt.plot(dates_test, y_test, label="Températures observées")
plt.plot(dates_test, predictions, label="Températures prédites")
plt.title("NEXORA SCIENCE — Prédictions expérimentales")
plt.xlabel("Date")
plt.ylabel("Température (°C)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from datetime import datetime

# NEXORA SCIENCE — Démonstration autonome

donnees = pd.DataFrame({
    "Jour": range(1, 8),
    "Temperature": [25.2, 26.1, 24.8, 27.0, 26.4, 25.9, 27.3]
})

print("NEXORA SCIENCE — ANALYSE SCIENTIFIQUE")
print("=" * 40)

print("\n1. Données analysées")
display(donnees)

print("\n2. Résultats")
moyenne = donnees["Temperature"].mean()
minimum = donnees["Temperature"].min()
maximum = donnees["Temperature"].max()

print("Moyenne :", round(moyenne, 2), "°C")
print("Minimum :", minimum, "°C")
print("Maximum :", maximum, "°C")

print("\n3. Graphique")
plt.figure(figsize=(8, 4))
plt.plot(donnees["Jour"], donnees["Temperature"], marker="o")
plt.title("NEXORA SCIENCE — Températures")
plt.xlabel("Jour")
plt.ylabel("Température (°C)")
plt.grid(True)
plt.show()

print("\n4. Rapport")
rapport = (
    "NEXORA SCIENCE\n"
    "RAPPORT D'ANALYSE\n"
    f"Date : {datetime.now():%d/%m/%Y %H:%M}\n"
    f"Moyenne : {moyenne:.2f} °C\n"
    f"Minimum : {minimum:.2f} °C\n"
    f"Maximum : {maximum:.2f} °C\n"
    "Note : démonstration avec des données fictives."
)

with open("rapport_nexora_science.txt", "w", encoding="utf-8") as fichier:
    fichier.write(rapport)

print(rapport)
print("\nRapport créé avec succès.")
```

```python
import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
import tempfile

def analyser_csv(fichier):
    if fichier is None:
        return "Importe un fichier CSV.", None, None

    try:
        df = pd.read_csv(fichier)

        if df.empty:
            return "Le fichier est vide.", None, None

        numeriques = df.select_dtypes(include="number")

        rapport = (
            "NEXORA SCIENCE — RAPPORT SCIENTIFIQUE\n"
            "======================================\n\n"
            f"Nombre de lignes : {len(df)}\n"
            f"Nombre de colonnes : {len(df.columns)}\n"
            f"Valeurs manquantes : {df.isna().sum().sum()}\n\n"
            "STATISTIQUES\n"
            + (
                numeriques.describe().round(2).to_string()
                if not numeriques.empty
                else "Aucune colonne numérique détectée."
            )
        )

        graphique = None

        if not numeriques.empty:
            fig, ax = plt.subplots(figsize=(8, 4))
            numeriques.iloc[:, 0].plot(ax=ax)
            ax.set_title("NEXORA SCIENCE — Analyse des données")
            ax.set_xlabel("Observation")
            ax.set_ylabel(str(numeriques.columns[0]))
            ax.grid(True)
            fig.tight_layout()
            graphique = fig

        fichier_rapport = tempfile.NamedTemporaryFile(
            mode="w",
            suffix=".txt",
            prefix="rapport_nexora_",
            encoding="utf-8",
            delete=False
        )

        fichier_rapport.write(rapport)
        fichier_rapport.close()

        return rapport, graphique, fichier_rapport.name

    except Exception as erreur:
        return f"Erreur : {erreur}", None, None

with gr.Blocks(title="NEXORA SCIENCE") as application:
    gr.Markdown("# 🔬 NEXORA SCIENCE")
    gr.Markdown("### Plateforme d'analyse scientifique")

    fichier = gr.File(
        label="Importer tes données (CSV)",
        file_types=[".csv"],
        type="filepath"
    )

    bouton = gr.Button("Analyser les données")
    resultat = gr.Textbox(label="Rapport scientifique", lines=15)
    graphique = gr.Plot(label="Graphique")
    telechargement = gr.File(label="Télécharger le rapport")

    bouton.click(
        analyser_csv,
        inputs=[fichier],
        outputs=[resultat, graphique, telechargement]
    )

application.launch()
```
